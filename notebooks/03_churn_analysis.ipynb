```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/processed/clean_customer_churn.csv")
```

```python
overall_churn_rate = df["churn_value"].mean()
overall_churn_rate
```

```python
churn_by_contract = df.groupby("contract")["churn_value"].mean()
churn_by_tenure = df.groupby("tenure_bucket")["churn_value"].mean()
churn_by_revenue = df.groupby("monthly_revenue_segment")["churn_value"].mean()
```

```python
avg_lifespan = df.groupby("churn_label")["tenure_months"].mean()
avg_revenue = df.groupby("churn_label")["monthly_charges"].mean()

ltv = avg_lifespan * avg_revenue
ltv
```
